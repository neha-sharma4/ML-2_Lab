```python
def find_s(data):
    n = len(data[0]) - 1

    hypothesis = ['0'] * n

    for example in data:

        attributes = example[:-1]
        target = example[-1]

        if target == "Yes":

            if hypothesis == ['0'] * n:
                hypothesis = attributes.copy()

            else:
                for i in range(n):
                    if hypothesis[i] != attributes[i]:
                        hypothesis[i] = '?'

            print("Hypothesis:", hypothesis)

    return hypothesis


data = [
    ["Sunny", "Warm", "Normal", "Strong", "Warm", "Same", "Yes"],
    ["Sunny", "Warm", "High",   "Strong", "Warm", "Same", "Yes"],
    ["Rainy", "Cold", "High",   "Strong", "Warm", "Change", "No"],
    ["Sunny", "Warm", "High",   "Strong", "Cool", "Change", "Yes"]
]



final_hypothesis = find_s(data)



print("\nFinal Hypothesis:")
print(final_hypothesis)
```

```python
# Graph: Positive vs Negative Examples

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

columns = [
    "Sky", "AirTemp", "Humidity",
    "Wind", "Water", "Forecast", "EnjoySport"
]

df = pd.DataFrame(data, columns=columns)

# Bar graph
plt.figure(figsize=(6, 4))
df["EnjoySport"].value_counts().plot(kind="bar")
plt.title("Positive vs Negative Examples")
plt.xlabel("EnjoySport")
plt.ylabel("Number of Examples")
plt.xticks(rotation=0)
plt.show()


# Heatmap
encoded_df = df.copy()

for column in encoded_df.columns:
    encoded_df[column] = pd.factorize(encoded_df[column])[0]

plt.figure(figsize=(10, 5))
sns.heatmap(
    encoded_df,
    annot=True,
    cmap="YlGnBu",
    linewidths=0.5
)

plt.title("Heatmap of EnjoySport Dataset")
plt.xlabel("Attributes")
plt.ylabel("Examples")
plt.show()
```
